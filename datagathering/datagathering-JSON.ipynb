```python
import json
import pandas as pd
```

```python
def load_large_file(filepath , chunksize) :
  chunklist = []
  with open(filepath) as f :
    data = json.load(f)
    for i in range(0 , len(data) ,chunksize) :
      chunk = pd.DataFrame(data[i : i+chunksize])
      chunklist.append(chunk)
  return pd.concat(chunklist , axis =0)

```

```python
data = load_large_file('train.json' , 10000)
```

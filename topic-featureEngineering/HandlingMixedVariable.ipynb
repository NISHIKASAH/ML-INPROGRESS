```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('titanic.csv')
df.sample(10)
```

```python
df['number_numberical'] = pd.to_numeric(df['number'] , errors='coerce' , downcast='integer')

df['number_categorical'] = np.where(df['number_numberical'].isnull() , df['number'] , np.nan )
df.head()
```

```python
df['cabin_num'] = df['Cabin'].str.extract(r'(\d+)')

df['Cabin_categorial'] = df['Cabin'].str[0]
df.head()
```

```python
df['Ticket_numberical'] =  df['Ticket'].apply(lambda x : x.split()[-1])
df['Ticket_numberical']=  pd.to_numeric(df['Ticket_numberical'] , errors = 'coerce' , downcast = 'integer')
df['Ticket_categorical'] = df['Ticket'].apply(lambda x :x.split()[0])
df['Ticket_categorical'] = np.where(df['Ticket_categorical'].str.isdigit() , np.nan , df['Ticket_categorical'])
df.head()
```

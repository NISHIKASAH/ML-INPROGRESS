```python
##performing hotmodule using pandas get_dummies
#dropping first column to prevent multicollinearity( relationship among columns i.e one variable can be predicted by other) , impact model

df_encoded = pd.get_dummies(df  , columns=['fuel' , 'owner'] ,drop_first = True)
```

```python
df_encoded
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
```

```python
x_train , x_test , y_train  , y_test = train_test_split(df.iloc[ : , 0:4] , df.iloc[: ,-1] , test_size = 0.2 , random_state=2 )
```

```python
ohe = OneHotEncoder(drop= 'first' , dtype= np.int32)
x_train_encoded = ohe.fit_transform(x_train[['fuel' , 'owner']]).toarray()
x_test_encoded = ohe.transform(x_test[['fuel' , 'owner']]).toarray()
```

```python
## we splitted the brand and km_driven and then after encoding fuel and owner we merge using numpy hstack
x_train_final = np.hstack(( x_train[['brand' , 'km_driven']].values ,  x_train_encoded))
```

```python
x_test_final = np.hstack(( x_test[['brand' , 'km_driven']].values , x_test_encoded))
```

```python
print('Training data' , x_train_final.shape)
```

```python
print('Test data' , x_test_final.shape)
```

```python
## implementing cardinality - to group rare categories into 1 group ,reduces dimensions

brand_counts = df['brand'].value_counts()
```

```python
df['brand'].nunique()
```

```python
threshold = 100 
rare_brands = brand_counts[brand_counts <= threshold ].index
```

```python
df['brand'] = df['brand'].replace(rare_brands , "unknown") 
```
